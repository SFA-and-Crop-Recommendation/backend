```python
import os
import requests
import joblib
import pandas as pd
from datetime import datetime, timedelta
from prophet import Prophet

API_KEY = "579b464db66ec23bdd00000107d05e8dc0f44b2264e86594199d6d7f"
API_URL = "https://api.data.gov.in/resource/35985678-0d79-46b4-9ed6-6f13308a1d24"
MODEL_DIR = "models"
os.makedirs(MODEL_DIR, exist_ok=True)

def fetch_city_crop_data(city, crop, months_back=24):
    end_date = datetime.today()
    start_date = end_date - timedelta(days=30 * months_back)
    records = []
    offset = 0

    while True:
        params = {
            "api-key": API_KEY,
            "format": "json",
            "filters[Commodity]": crop,
            "filters[market]": city,
            "offset": offset,
            "limit": 1000
        }
        response = requests.get(API_URL, params=params)
        if response.status_code != 200:
            print(f"⚠️ API error for {city}-{crop}")
            break

        data = response.json().get("records", [])
        if not data:
            break

        for row in data:
            try:
                date = datetime.strptime(row["arrival_date"], "%Y-%m-%d")
                if date >= start_date:
                    records.append({"ds": date, "y": float(row["modal_price"])})
            except:
                continue
        offset += 1000

    df = pd.DataFrame(records)
    if df.empty:
        return None
    df = df.groupby("ds")["y"].mean().reset_index()
    return df.sort_values("ds")

def get_model_filename(city, crop):
    return os.path.join(MODEL_DIR, f"{city.lower()}_{crop.lower()}_prophet.pkl".replace(" ", "_"))

def train_model_if_needed(city, crop):
    model_path = get_model_filename(city, crop)
    if os.path.exists(model_path):
        print(f"📂 Found existing model: {model_path}")
        return joblib.load(model_path)

    print(f"📡 No model found. Training model for {city}-{crop}...")
    df = fetch_city_crop_data(city, crop)
    if df is None or df.empty:
        print(f"❌ No data found for {city}-{crop}.")
        return None

    model = Prophet()
    model.fit(df)
    joblib.dump(model, model_path)
    print(f"✅ Trained and saved model to: {model_path}")
    return model

def predict_price(city, crop, days=180):
    model = train_model_if_needed(city, crop)
    if model is None:
        return

    future = model.make_future_dataframe(periods=days)
    forecast = model.predict(future)
    result = forecast[["ds", "yhat"]].tail(days)
    print(f"\n📈 Predicted Prices for '{crop}' in '{city}' (next {days} days):\n")
    print(result.to_string(index=False))

# ========== MAIN ==========
if __name__ == "__main__":
    print("🌾 Crop Price Predictor for Indian Markets")
    city = input("Enter City/Market (e.g., Mumbai): ").strip()
    crop = input("Enter Crop (e.g., Tomato): ").strip()

    if city and crop:
        predict_price(city, crop, days=180)
    else:
        print("❗ Please provide both city and crop name.")
```

```python
import os
import requests
import pandas as pd
from datetime import datetime, timedelta
from prophet import Prophet
import joblib

API_KEY = "579b464db66ec23bdd00000107d05e8dc0f44b2264e86594199d6d7f"
API_URL = "https://api.data.gov.in/resource/35985678-0d79-46b4-9ed6-6f13308a1d24"
MODEL_DIR = "models"
os.makedirs(MODEL_DIR, exist_ok=True)

def fetch_crop_data_from_api(city, crop, months=24):
    end_date = datetime.today()
    start_date = end_date - timedelta(days=30 * months)
    records, offset = [], 0

    while True:
        params = {
            "api-key": API_KEY,
            "format": "json",
            "filters[Commodity]": crop,
            "filters[market]": city,
            "offset": offset,
            "limit": 1000
        }
        response = requests.get(API_URL, params=params)
        if response.status_code != 200:
            print("❌ API request failed.")
            break

        data = response.json().get("records", [])
        if not data:
            break

        for row in data:
            try:
                date = datetime.strptime(row["arrival_date"], "%Y-%m-%d")
                if date >= start_date:
                    records.append({
                        "ds": date,
                        "y": float(row["modal_price"])
                    })
            except:
                continue
        offset += 1000

    df = pd.DataFrame(records)
    if df.empty:
        return None
    return df.groupby("ds")["y"].mean().reset_index()

def get_model_path(city, crop):
    return os.path.join(MODEL_DIR, f"{city.lower()}_{crop.lower()}.pkl".replace(" ", "_"))

def train_or_load_model(city, crop):
    model_path = get_model_path(city, crop)
    if os.path.exists(model_path):
        print(f"📂 Loading cached model for {city} - {crop}")
        return joblib.load(model_path)

    print(f"📡 Fetching data from API for {city} - {crop}...")
    df = fetch_crop_data_from_api(city, crop)
    if df is None or df.empty:
        print("❌ No data available.")
        return None

    print("🧠 Training Prophet model...")
    model = Prophet()
    model.fit(df)
    joblib.dump(model, model_path)
    print(f"✅ Model saved: {model_path}")
    return model

def predict_crop_price(city, crop, days=180):
    model = train_or_load_model(city, crop)
    if model is None:
        return

    future = model.make_future_dataframe(periods=days)
    forecast = model.predict(future)
    result = forecast[["ds", "yhat"]].tail(days)
    print(f"\n📈 Price Forecast for '{crop}' in '{city}' (Next {days} days):")
    print(result.to_string(index=False))

# ===== MAIN =====
if __name__ == "__main__":
    print("🌾 Crop Price Predictor (India API)")
    city = input("Enter market (city), e.g., 'Mumbai', 'Kolkata (North)': ").strip()
    crop = input("Enter crop name, e.g., 'Tomato', 'Potato': ").strip()

    if not city or not crop:
        print("❗ Please enter both city and crop.")
    else:
        predict_crop_price(city, crop)
```
