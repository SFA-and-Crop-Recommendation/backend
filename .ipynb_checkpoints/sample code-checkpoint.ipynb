```python
import pandas as pd
import json

# Assuming 'predict_next_6_months' is defined and working properly

df = pd.read_csv("crop_prices.csv")

# Define the prediction function
def forecast_to_json(crop, market, df):
    forecast = predict_next_6_months(crop, market, df)

    # Check if forecast is available
    if forecast:
        # Convert forecast dictionary to a JSON-compatible format, ensuring float32 is converted to standard float
        forecast_json = {str(date): float(price) for date, price in forecast.items()}
        return json.dumps(forecast_json, indent=2)
    else:
        return json.dumps({"error": "No forecast available for the given crop and market."}, indent=2)

# Example usage
result_json = forecast_to_json("Rice", "Sealdah Koley Market", df)
print(result_json)


```

```python
import pandas as pd
import json
import pickle
import numpy as np
import warnings

# Suppress specific scaler warning
warnings.filterwarnings("ignore", message="X does not have valid feature names")

# Load the model pipeline
with open(r'crop_recommendation_model.pkl', 'rb') as f:
    model_pipeline = pickle.load(f)

model = model_pipeline['model']
scaler = model_pipeline['scaler']
label_encoder = model_pipeline['label_encoder']
feature_columns = model_pipeline['feature_columns']

# Define your test parameters
test_params = [90, 42, 43, 20.87, 82.00, 6.5, 202.0]

# Prepare input sample for prediction
test_sample_df = pd.DataFrame([test_params], columns=feature_columns)
test_sample_scaled = scaler.transform(test_sample_df)

# Predict probabilities
pred_proba = model.predict_proba(test_sample_scaled)

# Recommend Top 5 Crops based on predicted probabilities
top_5_indices = np.argsort(pred_proba[0])[::-1][:5]
crops = label_encoder.inverse_transform(top_5_indices)
crops = crops.tolist()  # Convert numpy array to list

# Make sure df is loaded (replace with your actual data loading code if needed)
# Example: df = pd.read_csv("your_market_data.csv")
# df should contain 'District', 'Market', and relevant columns for the process.

# Get all Kolkata markets
kolkata_markets = df[df['District'].str.lower().str.contains("kolkata", na=False)]['Market'].unique()

# Collect results
all_results = []
crops=['rice', 'jute', 'coffee', 'banana', 'muskmelon']
# Loop over all markets in Kolkata
for market in kolkata_markets:
    results = crop_profit_recommendation_with_risk(crops, df, market)  # Use crops from model output
    all_results.extend(results)

# Convert results to DataFrame
result_df = pd.DataFrame(all_results)

# Ensure 'Market' exists in the result and drop duplicates
if 'Market' in result_df.columns:
    result_df.drop_duplicates(subset=["Crop", "Predicted_Price (₹/qtl)", "Market"], inplace=True)

# Ensure 'Net_Profit_per_Acre (₹)' exists before sorting
if 'Net_Profit_per_Acre (₹)' in result_df.columns:
    result_df = result_df.sort_values(by="Net_Profit_per_Acre (₹)", ascending=False).reset_index(drop=True)
else:
    print("❌ Column 'Net_Profit_per_Acre (₹)' not found in result_df")

# Get top 3 crops
top3 = result_df.head(3)

# Convert to JSON and print
top3_json = json.dumps(top3.to_dict(orient="records"), indent=2)
print(top3_json)
```

```python
import os
import pandas as pd
import numpy as np
import joblib
from tensorflow.keras.models import load_model

def crop_profit_recommendation_with_risk(df, crops, market):
    """
    Predicts prices using pretrained models and recommends top 3 crops based on profit and risk.
    """
    yield_per_acre = {
        "Wheat": 20,
        "Rice": 22,
        "Maize": 18,
        "Sugarcane": 400,
        "Potato": 250,
        "Tomato": 180,
        "Onion": 120,
        "Mustard": 10,
        "Cotton": 8,
        "Apple": 7,
        "Banana": 35
    }

    cost_per_acre = {
        "Wheat": 25000,
        "Rice": 30000,
        "Maize": 20000,
        "Sugarcane": 50000,
        "Potato": 60000,
        "Tomato": 40000,
        "Onion": 35000,
        "Mustard": 18000,
        "Cotton": 45000,
        "Apple": 90000,
        "Banana": 50000
    }

    risk_factors = {
        "Wheat": "🟢 Low",
        "Rice": "🟢 Low",
        "Maize": "🟢 Low",
        "Sugarcane": "🟡 Medium",
        "Potato": "🔴 High",
        "Tomato": "🔴 High",
        "Onion": "🔴 High",
        "Mustard": "🟢 Low",
        "Cotton": "🟡 Medium",
        "Apple": "🔴 High",
        "Banana": "🔴 High"
    }

    results = []

    for crop in crops:
        try:
            # Define paths
            model_path = f"lstm_models/lstm_{crop}_{market}.h5"
            scaler_path = f"scalers/scaler_{crop}_{market}.pkl"

            # Debugging - print the paths
            print(f"Checking model at: {model_path}")
            print(f"Checking scaler at: {scaler_path}")

            # # Check if the model and scaler files exist
            # if not os.path.exists(model_path):
            #     print(f"❌ Model file not found: {model_path}")
            #     continue
            # if not os.path.exists(scaler_path):
            #     print(f"❌ Scaler file not found: {scaler_path}")
            #     continue

            # Load model and scaler
            model = load_model(model_path, compile=False)
            scaler = joblib.load(scaler_path)

            # Filter crop data
            data = df[(df['Commodity'] == crop) & (df['Market'] == market)][['Arrival_Date', 'Modal_Price']].copy()
            data['Arrival_Date'] = pd.to_datetime(data['Arrival_Date'], dayfirst=True, errors='coerce')
            data.dropna(inplace=True)
            data.set_index('Arrival_Date', inplace=True)
            data = data.resample("ME").mean().dropna()

            if len(data) < 12:
                continue

            last_12 = data[-12:].values
            last_scaled = scaler.transform(last_12).reshape(1, 12, 1)
            forecast = model.predict(last_scaled, verbose=0)[0]
            price = float(scaler.inverse_transform(forecast.reshape(-1, 1)).flatten()[0])

            yield_qtl = yield_per_acre.get(crop)
            cost = cost_per_acre.get(crop)
            risk = risk_factors.get(crop, "🟡 Medium")

            if yield_qtl is None or cost is None:
                continue

            revenue = price * yield_qtl
            profit = revenue - cost

            results.append({
                "Crop": crop,
                "Predicted_Price (₹/qtl)": round(price, 2),
                "Yield_per_Acre (qtl)": yield_qtl,
                "Revenue_per_Acre (₹)": round(revenue, 2),
                "Cost_per_Acre (₹)": cost,
                "Net_Profit_per_Acre (₹)": round(profit, 2),
                "Risk_Factor": risk
            })

        except Exception as e:
            print(f"❌ Error for {crop}: {e}")
            continue

    results = sorted(results, key=lambda x: x["Net_Profit_per_Acre (₹)"], reverse=True)
    
    return results[:3] if results else [{"message": "No valid crop predictions available."}]
```

```python
# Example DataFrame and market (you can replace this with actual data)
import pandas as pd

# Assuming df is already loaded with necessary market data
# Example of what df might look like
# df = pd.read_csv("your_data.csv")

# Example crops list (you can use crops from the model output or predefined list)
crops = ["Wheat", "Rice", "Potato", "Maize", "Sugarcane"]

# Example market (you can replace this with actual market names)
market = "Sealdah Koley Market"

# Call the crop_profit_recommendation_with_risk function
recommendations = crop_profit_recommendation_with_risk(df, crops, market)

# Print the recommendations
print(recommendations)
```
